```python
from sklearn import svm, datasets
iris = datasets.load_iris()
```

```python
import pandas as pd
df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
df['flower'] = iris.target
df['flower'] = df['flower'].apply(lambda x: iris.target_names[x])
df[47:52]
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(iris.data, iris.target, test_size=0.3)
```

```python
model = svm.SVC(kernel='rbf', C=3, gamma='auto')
model.fit(X_train, y_train)
model.score(X_test, y_test)
```

```python
from sklearn.model_selection import cross_val_score
```

```python
cross_val_score(svm.SVC(kernel='linear', C=10, gamma='auto'), iris.data, iris.target, cv=5)
```

```python
cross_val_score(svm.SVC(kernel='rbf', C=10, gamma='auto'), iris.data, iris.target, cv=5)
```

```python
cross_val_score(svm.SVC(kernel='rbf', C=20, gamma='auto'), iris.data, iris.target, cv=5)
```

```python
Kernels = ['linear', 'poly', 'rbf', 'sigmoid']
C = [1, 10, 20, 30, 40, 50]
avg_score = {}
for kval in Kernels:
    for cval in C:
        cv_score = cross_val_score(
            svm.SVC(kernel=kval, C=cval, gamma='auto'),
            iris.data,
            iris.target,
            cv=5
        )
        avg_score[f'{kval}_{cval}'] = cv_score.mean()

avg_scores = avg_score
avg_scores
```

Using Grid search CV


```python
from sklearn.model_selection import GridSearchCV

clf = GridSearchCV(svm.SVC(gamma='auto'), {
    'C': [1, 10, 20, 30, 40, 50],
    'kernel': ['linear', 'poly', 'rbf', 'sigmoid']
}, cv=5, return_train_score=False)

clf.fit(iris.data, iris.target)
clf.cv_results_
```

```python
df = pd.DataFrame(clf.cv_results_)
df
```

```python
df[['param_C', 'param_kernel', 'mean_test_score', 'std_test_score', 'rank_test_score']].sort_values(by='rank_test_score')
```

```python
clf.best_score_
```

```python
clf.best_params_
```

```python
from sklearn.model_selection import RandomizedSearchCV
rs = RandomizedSearchCV(svm.SVC(gamma='auto'), {
    'C': [1, 10, 20, 30, 40, 50],
    'kernel': ['linear', 'poly', 'rbf', 'sigmoid']}, cv=5, return_train_score=False, n_iter=2)
rs.fit(iris.data, iris.target)
pd.DataFrame(rs.cv_results_)[['param_C', 'param_kernel', 'mean_test_score', 'std_test_score', 'rank_test_score']].sort_values(by='rank_test_score')
```

choosing the best model

```python
from sklearn import svm
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
```

```python
model_params = {
    'svm': {
        'model': svm.SVC(gamma='auto'),
        'params': {
            'C': [1, 10, 20],
            'kernel': ['rbf', 'linear']
        }
    },
    'random_forest': {
        'model': RandomForestClassifier(),
        'params': {
            'n_estimators': [1, 5, 10]
        }
    },
    'logistic_regression': {
        'model': LogisticRegression(solver='liblinear', max_iter=1000),
        'params': {
            'C': [1, 5, 10]
        }
    }
}
```

```python
scores = []
for model_name, mp in model_params.items():
    clf = GridSearchCV(mp['model'], mp['params'], cv=5, return_train_score=False)
    if model_name == 'logistic_regression':
        clf.estimator.set_params(solver='lbfgs')

    clf.fit(iris.data, iris.target)
    scores.append({
        'model': model_name,
        'best_score': clf.best_score_,
        'best_params': clf.best_params_
    })
```

```python
df = pd.DataFrame(scores, columns=['model', 'best_score', 'best_params'])
df
```
