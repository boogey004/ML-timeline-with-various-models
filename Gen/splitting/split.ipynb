Spliting of data with sklearn into train and test

Importing Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

Importing Data

```python
df = pd.read_csv('car_prices_practice.csv')
df
```

Exploring data

```python
df.duplicated().sum()
```

```python
df.duplicated()
```

```python
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
df.isnull().sum()
```

```python
plt.scatter(df['Mileage'], df['sell price'])
```

```python
plt.scatter(df['age'], df['sell price'])
```

Declaring X & y. Splitting data into test and train

```python
X = df[['Mileage', 'age']]
y = df['sell price']
```

```python
X
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

Importing, training and testing model

```python
from sklearn.linear_model import LinearRegression
clf = LinearRegression()
```

```python
clf.fit(X_train, y_train)
```

```python
clf.predict(X_test)
```

```python
y_test
```

```python
clf.score(X_test, y_test)
```

```python
df_test = pd.read_csv('car_prices_test_set.csv')
df_test
```

```python
X_test = df_test.drop(columns=["sell price"])
y_test = df_test.drop(columns=['Mileage', 'age'])
```

```python
clf.predict(X_test)
```

```python
clf.score(X_test, y_test)
```
