Working with categorical data and organizing them using one hot encoder

Importing libraries

```python
import pandas as pd
```

```python
df = pd.read_csv("housing_categorical_practice.csv")
df
```

```python
dummies = pd.get_dummies(df['town'], dtype= int)
dummies
```

```python
merged = pd.concat([df,dummies], axis = 'columns')
merged
```

```python
final = merged.drop(['town', 'Oakwood'], axis='columns')
final
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
```

```python
X = final.drop('price', axis='columns')
X
```

```python
y = final.price
y
```

```python
model.fit(X,y)
```

```python
model.predict(pd.DataFrame(
    [[2800, 1, 0]],
    columns=X.columns
))
```

```python
model.score(X,y)
```
