Multi-Linear Regression

Importing Libraries

```python
import pandas as pd
import numpy as np
from sklearn import linear_model
```

Importing data

```python
df = pd.read_csv('house_prices.csv')
df
```

Exploring Data

```python
df.shape
```

```python
df.duplicated().sum()
```

```python
df.isnull().sum()
```

```python
df.describe()
```

Fixing missing data using median

```python
import math
median_bedrooms = math.floor(df.bedrooms.median())
median_bedrooms
```

```python
df.bedrooms = df.bedrooms.fillna(median_bedrooms)
df
```

```python
df.isnull().sum()
```

Using Linear Regression 

```python
reg = linear_model.LinearRegression()
reg.fit(df[['area', 'bedrooms', 'age']], df.price)
```

```python
reg.coef_
```

```python
reg.intercept_
```

```python
reg.predict([[3000,3,40]])
```
