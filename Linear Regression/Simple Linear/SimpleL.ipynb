Simple Linear REgression

Importing Libraries 

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn import linear_model
```

Importing Random Data Generated by Gemini

```python
house_df = pd.read_csv("house_prices_linear_regression.csv")
house_df
```

Eploring data

```python
%matplotlib inline
plt.xlabel('area')
plt.ylabel('price')
plt.scatter(house_df.area, house_df.price, color='red', marker='+')
```

```python
house_df.shape
```

```python
house_df.describe()
```

```python
house_df.duplicated().sum()
```

```python
house_df.isnull().sum()
```

Using linear Regression

```python
reg = linear_model.LinearRegression()
reg.fit(house_df[['area']],house_df.price)
```

```python
val = np.array([5000])
pred = reg.predict(val.reshape(1, -1))
print(pred)
```

```python
reg.coef_
```

```python
reg.intercept_
```

```python

144.53179251*5000+53635.95172019891
```

Visual repretensation of the regression line

```python
%matplotlib inline
plt.xlabel('area', fontsize=20)
plt.ylabel('price', fontsize=20)
plt.scatter(house_df.area, house_df.price, color='red', marker='+')
plt.plot(house_df.area, reg.predict(house_df[['area']]), color='blue')
```
